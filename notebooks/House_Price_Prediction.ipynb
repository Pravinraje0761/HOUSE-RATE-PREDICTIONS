## Step 2: Handle Missing Values

**Why handle missing values?**
- Machine learning algorithms cannot work with missing (NaN) values
- Missing values can lead to biased or incorrect predictions
- We need to either remove them or fill them with appropriate values

**Strategies for handling missing values:**
- **Numerical columns**: Fill with mean, median, or mode
- **Categorical columns**: Fill with mode or a special category
- **Rows with many missing values**: Remove the entire row

We'll use the following strategy:
- Numerical columns: Fill with median (robust to outliers)
- Categorical columns: Fill with mode (most frequent value)

```python
# Check missing values before handling
print("Missing values before handling:")
print(df.isnull().sum())
print(f"\nTotal missing values: {df.isnull().sum().sum()}")
```

```python
# Handle missing values for numerical columns (fill with median)
numerical_cols_with_missing = ['Area', 'Number_of_Bedrooms', 'Number_of_Bathrooms', 
                                'Parking', 'Age_of_House', 'Distance_to_City_Center', 'Nearby_Schools']

for col in numerical_cols_with_missing:
    if df[col].isnull().sum() > 0:
        median_value = df[col].median()
        df[col].fillna(median_value, inplace=True)
        print(f"Filled {col} with median value: {median_value}")

# Handle missing values for categorical columns (fill with mode)
categorical_cols_with_missing = ['Furnishing_Status']

for col in categorical_cols_with_missing:
    if df[col].isnull().sum() > 0:
        mode_value = df[col].mode()[0]
        df[col].fillna(mode_value, inplace=True)
        print(f"Filled {col} with mode value: {mode_value}")
```

```python
# Verify missing values after handling
print("Missing values after handling:")
print(df.isnull().sum())
print(f"\nTotal missing values: {df.isnull().sum().sum()}")
```

## Step 3: Detect and Remove Outliers

**Why handle outliers?**
- Outliers are extreme values that differ significantly from other observations
- They can skew statistical measures and affect model performance
- They may represent data entry errors or genuine rare cases

**Methods for outlier detection:**
- **IQR (Interquartile Range) Method**: Values below Q1 - 1.5*IQR or above Q3 + 1.5*IQR
- **Z-Score Method**: Values with z-score > 3 or < -3
- **Visualization**: Box plots to visually identify outliers

We'll use the IQR method for numerical columns.

```python
# Function to remove outliers using IQR method
def remove_outliers_iqr(df, column):
    """
    Remove outliers from a column using IQR method.
    
    Parameters:
    df (DataFrame): The dataframe
    column (str): The column name
    
    Returns:
    DataFrame: Dataframe with outliers removed
    """
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    # Filter data
    df_filtered = df[(df[column] >= lower_bound) & (df[column] <= upper_bound)]
    
    removed = len(df) - len(df_filtered)
    print(f"Removed {removed} outliers from {column}")
    
    return df_filtered

# Remove outliers from key numerical columns
print("Removing outliers using IQR method:")
print("=" * 50)

# We'll remove outliers from Price and Area as they are most important
df = remove_outliers_iqr(df, 'Price')
df = remove_outliers_iqr(df, 'Area')

print(f"\nDataset shape after outlier removal: {df.shape}")
```

## Step 4: Encode Categorical Variables

**Why encode categorical variables?**
- Machine learning algorithms work with numerical data only
- Categorical variables (text) need to be converted to numbers
- Different encoding methods for different types of categorical data

**Encoding methods:**
- **Label Encoding**: Convert categories to numbers (0, 1, 2, ...) - good for ordinal data
- **One-Hot Encoding**: Create binary columns for each category - good for nominal data

We'll use:
- Label Encoding for ordinal variables (Furnishing_Status, Crime_Rate)
- One-Hot Encoding for nominal variables (Location, City, Property_Type)

```python
# Initialize Label Encoder
label_encoder = LabelEncoder()

# Label Encoding for ordinal categorical variables
ordinal_cols = ['Furnishing_Status', 'Crime_Rate']

print("Label Encoding for ordinal variables:")
print("=" * 50)

for col in ordinal_cols:
    df[col + '_Encoded'] = label_encoder.fit_transform(df[col])
    print(f"{col}: {dict(zip(label_encoder.classes_, label_encoder.transform(label_encoder.classes_)))}")

# One-Hot Encoding for nominal categorical variables
nominal_cols = ['Location', 'City', 'Property_Type']

print("\nOne-Hot Encoding for nominal variables:")
print("=" * 50)

df = pd.get_dummies(df, columns=nominal_cols, drop_first=True, prefix=nominal_cols)

print(f"Shape after One-Hot Encoding: {df.shape}")
print(f"New columns created: {df.shape[1] - 16}")
```

## Step 5: Feature Scaling

**Why scale features?**
- Features have different units and scales (e.g., Area in sq ft, Age in years)
- ML algorithms (especially distance-based) can be biased towards features with larger scales
- Scaling ensures all features contribute equally to the model

**Scaling methods:**
- **StandardScaler**: Scales to mean=0, std=1 (good for normally distributed data)
- **MinMaxScaler**: Scales to range [0, 1] (good for bounded data)
- **RobustScaler**: Uses median and IQR (robust to outliers)

We'll use StandardScaler for numerical features.

```python
# Identify numerical columns to scale (excluding encoded categorical and target variable)
numerical_cols_to_scale = ['Area', 'Number_of_Bedrooms', 'Number_of_Bathrooms', 
                           'Number_of_Floors', 'Parking', 'Age_of_House',
                           'Distance_to_City_Center', 'Nearby_Schools', 'Nearby_Hospitals']

print("Scaling numerical features:")
print("=" * 50)

# Initialize StandardScaler
scaler = StandardScaler()

# Fit and transform the numerical columns
df[numerical_cols_to_scale] = scaler.fit_transform(df[numerical_cols_to_scale])

print("Numerical features scaled using StandardScaler")
print(f"Mean of scaled features (should be ~0): {df[numerical_cols_to_scale].mean().mean():.6f}")
print(f"Std of scaled features (should be ~1): {df[numerical_cols_to_scale].std().mean():.6f}")

# Display first few rows of scaled data
print("\nFirst 5 rows after scaling:")
print(df[numerical_cols_to_scale].head())
```

## Step 6: Drop Unnecessary Columns

We'll drop columns that are not needed for modeling:
- House_ID: Just an identifier, not useful for prediction
- Original categorical columns (after encoding): We have encoded versions

```python
# Drop unnecessary columns
columns_to_drop = ['House_ID', 'Furnishing_Status', 'Crime_Rate']

for col in columns_to_drop:
    if col in df.columns:
        df.drop(col, axis=1, inplace=True)
        print(f"Dropped column: {col}")

print(f"\nFinal dataset shape: {df.shape}")
print(f"Final columns: {list(df.columns)}")
```

# Part 4: Feature Engineering

Feature engineering creates new features from existing data to improve model performance. It helps:
- Capture complex relationships
- Improve model accuracy
- Reduce overfitting
- Make models more interpretable

---

## Feature 1: Price per Square Foot

**Why this feature?**
- Price per sq ft is a standard real estate metric
- It normalizes price by area, allowing better comparison
- It captures the value of location and amenities independent of size

```python
# Feature 1: Price per Square Foot
# We need to use the original (unscaled) area values
# First, let's get the original area values from the scaler
original_area = scaler.inverse_transform(df[numerical_cols_to_scale])[:, 0]  # Area is first column

df['Price_per_SqFt'] = df['Price'] / original_area

print("Feature 1: Price per Square Foot created")
print(f"Mean Price per SqFt: ${df['Price_per_SqFt'].mean():.2f}")
print(f"Range: ${df['Price_per_SqFt'].min():.2f} - ${df['Price_per_SqFt'].max():.2f}")
```

## Feature 2: Total Rooms

**Why this feature?**
- Combines bedrooms and bathrooms into a single metric
- Represents the overall size/capacity of the house
- Often used in real estate listings

```python
# Feature 2: Total Rooms
# Get original values for bedrooms and bathrooms
original_bedrooms = scaler.inverse_transform(df[numerical_cols_to_scale])[:, 1]
original_bathrooms = scaler.inverse_transform(df[numerical_cols_to_scale])[:, 2]

df['Total_Rooms'] = original_bedrooms + original_bathrooms

print("Feature 2: Total Rooms created")
print(f"Mean Total Rooms: {df['Total_Rooms'].mean():.2f}")
print(f"Range: {df['Total_Rooms'].min():.0f} - {df['Total_Rooms'].max():.0f}")
```

## Feature 3: Luxury Score

**Why this feature?**
- Combines multiple luxury indicators into a single score
- Higher score indicates more luxurious property
- Helps the model understand premium vs. standard properties

```python
# Feature 3: Luxury Score
# Based on furnishing status, parking, and property type
# Higher score = more luxurious

# Get original values
original_parking = scaler.inverse_transform(df[numerical_cols_to_scale])[:, 4]
original_floors = scaler.inverse_transform(df[numerical_cols_to_scale])[:, 3]

# Calculate luxury score (0-100 scale)
df['Luxury_Score'] = (
    df['Furnishing_Status_Encoded'] * 25 +  # Furnishing: 0-50
    original_parking * 10 +  # Parking: 0-30
    original_floors * 5 +  # Floors: 0-15
    (original_area / 100)  # Area contribution
)

# Normalize to 0-100 range
df['Luxury_Score'] = ((df['Luxury_Score'] - df['Luxury_Score'].min()) / 
                      (df['Luxury_Score'].max() - df['Luxury_Score'].min())) * 100

print("Feature 3: Luxury Score created")
print(f"Mean Luxury Score: {df['Luxury_Score'].mean():.2f}")
print(f"Range: {df['Luxury_Score'].min():.2f} - {df['Luxury_Score'].max():.2f}")
```

## Feature 4: Accessibility Score

**Why this feature?**
- Combines distance to city center, nearby schools, and hospitals
- Higher score indicates better accessibility to amenities
- Important for families and convenience

```python
# Feature 4: Accessibility Score
# Based on distance to city center, nearby schools, and hospitals
original_distance = scaler.inverse_transform(df[numerical_cols_to_scale])[:, 6]
original_schools = scaler.inverse_transform(df[numerical_cols_to_scale])[:, 7]
original_hospitals = scaler.inverse_transform(df[numerical_cols_to_scale])[:, 8]

# Calculate accessibility score (0-100 scale)
# Closer to city center = higher score
# More schools/hospitals = higher score
df['Accessibility_Score'] = (
    (50 - original_distance) * 2 +  # Distance: closer is better
    original_schools * 5 +  # Schools: more is better
    original_hospitals * 5  # Hospitals: more is better
)

# Normalize to 0-100 range
df['Accessibility_Score'] = ((df['Accessibility_Score'] - df['Accessibility_Score'].min()) / 
                             (df['Accessibility_Score'].max() - df['Accessibility_Score'].min())) * 100

print("Feature 4: Accessibility Score created")
print(f"Mean Accessibility Score: {df['Accessibility_Score'].mean():.2f}")
print(f"Range: {df['Accessibility_Score'].min():.2f} - {df['Accessibility_Score'].max():.2f}")
```

## Feature 5: House Age Category

**Why this feature?**
- Categorizes houses into age groups (New, Modern, Old, Historic)
- Helps the model understand age-related price patterns
- More interpretable than continuous age

```python
# Feature 5: House Age Category
original_age = scaler.inverse_transform(df[numerical_cols_to_scale])[:, 5]

# Create age categories
def categorize_age(age):
    if age <= 5:
        return 'New'
    elif age <= 15:
        return 'Modern'
    elif age <= 30:
        return 'Old'
    else:
        return 'Historic'

df['House_Age_Category'] = original_age.apply(categorize_age)

# Encode the age category
age_category_encoder = LabelEncoder()
df['House_Age_Category_Encoded'] = age_category_encoder.fit_transform(df['House_Age_Category'])

print("Feature 5: House Age Category created")
print(f"Age Categories: {dict(zip(age_category_encoder.classes_, age_category_encoder.transform(age_category_encoder.classes_)))}")
print(f"\nDistribution:")
print(df['House_Age_Category'].value_counts())
```

## Feature 6: Location Popularity Score

**Why this feature?**
- Combines location type with crime rate
- Higher score indicates more desirable location
- Captures the desirability factor of location

```python
# Feature 6: Location Popularity Score
# Based on location type and crime rate
# Lower crime rate = higher score
# Certain locations (Downtown, Coastal) = higher score

# Get location columns from one-hot encoding
location_cols = [col for col in df.columns if col.startswith('Location_')]
crime_encoded = df['Crime_Rate_Encoded']

# Calculate location popularity score
location_score = 0
for col in location_cols:
    if 'Downtown' in col or 'Coastal' in col:
        location_score += df[col] * 30
    elif 'Suburban' in col or 'Urban' in col:
        location_score += df[col] * 20
    else:
        location_score += df[col] * 10

# Crime penalty: Low crime (0) = no penalty, High crime (2) = high penalty
crime_penalty = crime_encoded * 20

df['Location_Popularity_Score'] = location_score - crime_penalty

# Normalize to 0-100 range
df['Location_Popularity_Score'] = ((df['Location_Popularity_Score'] - df['Location_Popularity_Score'].min()) / 
                                    (df['Location_Popularity_Score'].max() - df['Location_Popularity_Score'].min())) * 100

print("Feature 6: Location Popularity Score created")
print(f"Mean Location Popularity Score: {df['Location_Popularity_Score'].mean():.2f}")
print(f"Range: {df['Location_Popularity_Score'].min():.2f} - {df['Location_Popularity_Score'].max():.2f}")
```

## Step 2: Remove Highly Correlated Features

Features with very high correlation (>0.9) with each other can cause multicollinearity. We'll identify and remove such features.

```python
# Find highly correlated features
correlation_matrix = df.corr()
high_corr_pairs = []

# Iterate through the correlation matrix
for i in range(len(correlation_matrix.columns)):
    for j in range(i+1, len(correlation_matrix.columns)):
        if abs(correlation_matrix.iloc[i, j]) > 0.9:
            high_corr_pairs.append((correlation_matrix.columns[i], 
                                   correlation_matrix.columns[j], 
                                   correlation_matrix.iloc[i, j]))

if high_corr_pairs:
    print("Highly correlated feature pairs (correlation > 0.9):")
    print("=" * 60)
    for pair in high_corr_pairs:
        print(f"{pair[0]} - {pair[1]}: {pair[2]:.3f}")
else:
    print("No highly correlated feature pairs found (correlation > 0.9)")
    print("All features can be retained for modeling")
```

## Step 3: Select Final Features

Based on correlation analysis, we'll select features with significant correlation to Price. We'll keep features with |correlation| > 0.1.

```python
# Select features with significant correlation to Price
threshold = 0.1
selected_features = correlation_with_price[abs(correlation_with_price) > threshold].index.tolist()
selected_features.remove('Price')  # Remove target variable

print(f"Selected features (|correlation| > {threshold}):")
print("=" * 60)
print(f"Number of features selected: {len(selected_features)}")
print("\nSelected features:")
for i, feature in enumerate(selected_features, 1):
    corr = correlation_with_price[feature]
    print(f"{i}. {feature}: {corr:.3f}")

# Create final dataset with selected features
df_final = df[selected_features + ['Price']].copy()
print(f"\nFinal dataset shape: {df_final.shape}")
```

# Part 6: Train-Test Split

Train-test split divides the dataset into training and testing sets. This is crucial because:
- **Training set**: Used to train the model (80%)
- **Testing set**: Used to evaluate model performance (20%)
- **Prevents overfitting**: Model is evaluated on unseen data
- **Random state**: Ensures reproducibility

---

## Split the Data

We'll split the data into features (X) and target (y), then into training and testing sets.

# Part 7: Machine Learning Models

In this section, we'll train and compare three regression models:
1. **Linear Regression**: Simple, interpretable, assumes linear relationship
2. **Decision Tree Regressor**: Non-linear, captures complex patterns
3. **Random Forest Regressor**: Ensemble method, combines multiple decision trees

We'll evaluate each model using:
- **R² Score**: Proportion of variance explained (0-1, higher is better)
- **MAE (Mean Absolute Error)**: Average absolute difference
- **MSE (Mean Squared Error)**: Average squared difference
- **RMSE (Root Mean Squared Error)**: Square root of MSE

---

## Model 1: Linear Regression

Linear Regression is the simplest regression algorithm. It assumes a linear relationship between features and target.

**Advantages:**
- Simple and interpretable
- Fast to train
- Works well for linear relationships

**Disadvantages:**
- Assumes linear relationship
- Sensitive to outliers
- Can't capture complex patterns

```python
# Initialize and train Linear Regression
lr_model = LinearRegression()
lr_model.fit(X_train, y_train)

# Make predictions
lr_train_pred = lr_model.predict(X_train)
lr_test_pred = lr_model.predict(X_test)

# Evaluate the model
lr_r2 = r2_score(y_test, lr_test_pred)
lr_mae = mean_absolute_error(y_test, lr_test_pred)
lr_mse = mean_squared_error(y_test, lr_test_pred)
lr_rmse = np.sqrt(lr_mse)

print("Linear Regression Results:")
print("=" * 60)
print(f"R² Score: {lr_r2:.4f}")
print(f"Mean Absolute Error (MAE): ${lr_mae:.2f}")
print(f"Mean Squared Error (MSE): ${lr_mse:.2f}")
print(f"Root Mean Squared Error (RMSE): ${lr_rmse:.2f}")
print(f"\nTraining R²: {r2_score(y_train, lr_train_pred):.4f}")
print(f"Testing R²: {lr_r2:.4f}")

# Check for overfitting
if r2_score(y_train, lr_train_pred) - lr_r2 > 0.1:
    print("\nWarning: Possible overfitting (Training R² much higher than Testing R²)")
else:
    print("\nNo significant overfitting detected")
```

## Model 2: Decision Tree Regressor

Decision Tree Regressor builds a tree-like model of decisions. It can capture non-linear relationships.

**Advantages:**
- Captures non-linear relationships
- Easy to interpret
- Handles both numerical and categorical data

**Disadvantages:**
- Prone to overfitting
- Can be unstable with small data changes
- May create complex trees

```python
# Initialize and train Decision Tree Regressor
dt_model = DecisionTreeRegressor(random_state=42)
dt_model.fit(X_train, y_train)

# Make predictions
dt_train_pred = dt_model.predict(X_train)
dt_test_pred = dt_model.predict(X_test)

# Evaluate the model
dt_r2 = r2_score(y_test, dt_test_pred)
dt_mae = mean_absolute_error(y_test, dt_test_pred)
dt_mse = mean_squared_error(y_test, dt_test_pred)
dt_rmse = np.sqrt(dt_mse)

print("Decision Tree Regressor Results:")
print("=" * 60)
print(f"R² Score: {dt_r2:.4f}")
print(f"Mean Absolute Error (MAE): ${dt_mae:.2f}")
print(f"Mean Squared Error (MSE): ${dt_mse:.2f}")
print(f"Root Mean Squared Error (RMSE): ${dt_rmse:.2f}")
print(f"\nTraining R²: {r2_score(y_train, dt_train_pred):.4f}")
print(f"Testing R²: {dt_r2:.4f}")

# Check for overfitting
if r2_score(y_train, dt_train_pred) - dt_r2 > 0.1:
    print("\nWarning: Possible overfitting (Training R² much higher than Testing R²)")
else:
    print("\nNo significant overfitting detected")
```

## Model 3: Random Forest Regressor

Random Forest is an ensemble method that combines multiple decision trees. It's one of the most powerful regression algorithms.

**Advantages:**
- High accuracy and performance
- Reduces overfitting through ensemble
- Handles large datasets well
- Provides feature importance

**Disadvantages:**
- More complex and less interpretable
- Slower to train
- Requires more memory

```python
# Initialize and train Random Forest Regressor
rf_model = RandomForestRegressor(n_estimators=100, random_state=42)
rf_model.fit(X_train, y_train)

# Make predictions
rf_train_pred = rf_model.predict(X_train)
rf_test_pred = rf_model.predict(X_test)

# Evaluate the model
rf_r2 = r2_score(y_test, rf_test_pred)
rf_mae = mean_absolute_error(y_test, rf_test_pred)
rf_mse = mean_squared_error(y_test, rf_test_pred)
rf_rmse = np.sqrt(rf_mse)

print("Random Forest Regressor Results:")
print("=" * 60)
print(f"R² Score: {rf_r2:.4f}")
print(f"Mean Absolute Error (MAE): ${rf_mae:.2f}")
print(f"Mean Squared Error (MSE): ${rf_mse:.2f}")
print(f"Root Mean Squared Error (RMSE): ${rf_rmse:.2f}")
print(f"\nTraining R²: {r2_score(y_train, rf_train_pred):.4f}")
print(f"Testing R²: {rf_r2:.4f}")

# Check for overfitting
if r2_score(y_train, rf_train_pred) - rf_r2 > 0.1:
    print("\nWarning: Possible overfitting (Training R² much higher than Testing R²)")
else:
    print("\nNo significant overfitting detected")
```

# Part 10: User Prediction System

In this section, we'll create an interactive prediction system that allows users to input house features and get a price prediction.

---

## Create a Prediction Function

We'll create a function that takes user input and returns a predicted house price.

```python
def predict_house_price(area, location, bedrooms, bathrooms, floors, parking, 
                        age, furnishing_status, distance_city, schools, hospitals, 
                        crime_rate, property_type, city):
    """
    Predict house price based on user input features.
    
    Parameters:
    - area: Area in square feet
    - location: Location type (Downtown, Suburban, Urban, Rural, Coastal)
    - bedrooms: Number of bedrooms
    - bathrooms: Number of bathrooms
    - floors: Number of floors
    - parking: Number of parking spaces
    - age: Age of house in years
    - furnishing_status: Furnishing status (Furnished, Semi-Furnished, Unfurnished)
    - distance_city: Distance to city center in km
    - schools: Number of nearby schools
    - hospitals: Number of nearby hospitals
    - crime_rate: Crime rate (Low, Medium, High)
    - property_type: Property type (Apartment, House, Villa, Condo, Townhouse)
    - city: City name
    
    Returns:
    - Predicted price
    """
    
    # Create a dictionary with input features
    input_data = {
        'Area': [area],
        'Number_of_Bedrooms': [bedrooms],
        'Number_of_Bathrooms': [bathrooms],
        'Number_of_Floors': [floors],
        'Parking': [parking],
        'Age_of_House': [age],
        'Distance_to_City_Center': [distance_city],
        'Nearby_Schools': [schools],
        'Nearby_Hospitals': [hospitals],
        'Furnishing_Status_Encoded': [0],  # Will be updated
        'Crime_Rate_Encoded': [0],  # Will be updated
    }
    
    # Encode categorical variables
    furnishing_map = {'Furnished': 2, 'Semi-Furnished': 1, 'Unfurnished': 0}
    crime_map = {'Low': 0, 'Medium': 1, 'High': 2}
    
    input_data['Furnishing_Status_Encoded'] = [furnishing_map.get(furnishing_status, 1)]
    input_data['Crime_Rate_Encoded'] = [crime_map.get(crime_rate, 1)]
    
    # Add one-hot encoded columns for location, city, and property type
    location_cols = [col for col in loaded_feature_names if col.startswith('Location_')]
    city_cols = [col for col in loaded_feature_names if col.startswith('City_')]
    property_cols = [col for col in loaded_feature_names if col.startswith('Property_Type_')]
    
    for col in location_cols:
        input_data[col] = [1 if col == f'Location_{location}' else 0]
    
    for col in city_cols:
        input_data[col] = [1 if col == f'City_{city}' else 0]
    
    for col in property_cols:
        input_data[col] = [1 if col == f'Property_Type_{property_type}' else 0]
    
    # Add engineered features
    input_data['Price_per_SqFt'] = [0]  # Placeholder
    input_data['Total_Rooms'] = [bedrooms + bathrooms]
    input_data['Luxury_Score'] = [50]  # Placeholder
    input_data['Accessibility_Score'] = [50]  # Placeholder
    input_data['House_Age_Category_Encoded'] = [1]  # Placeholder
    input_data['Location_Popularity_Score'] = [50]  # Placeholder
    
    # Create DataFrame
    input_df = pd.DataFrame(input_data)
    
    # Ensure all columns are present and in the right order
    for col in loaded_feature_names:
        if col not in input_df.columns:
            input_df[col] = 0
    
    input_df = input_df[loaded_feature_names]
    
    # Scale numerical features
    numerical_cols_to_scale = ['Area', 'Number_of_Bedrooms', 'Number_of_Bathrooms', 
                               'Number_of_Floors', 'Parking', 'Age_of_House',
                               'Distance_to_City_Center', 'Nearby_Schools', 'Nearby_Hospitals']
    
    input_df[numerical_cols_to_scale] = loaded_scaler.transform(input_df[numerical_cols_to_scale])
    
    # Make prediction
    prediction = loaded_model.predict(input_df)
    
    return prediction[0]

print("Prediction function created successfully!")
```

## Hyperparameter Tuning with GridSearchCV

GridSearchCV performs an exhaustive search over specified parameter values for an estimator.

```python
# Define parameter grid for Random Forest
param_grid = {
    'n_estimators': [50, 100, 200],
    'max_depth': [10, 20, None],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}

print("Starting GridSearchCV for hyperparameter tuning...")
print("This may take a few minutes...")

# Initialize GridSearchCV
grid_search = GridSearchCV(
    RandomForestRegressor(random_state=42),
    param_grid,
    cv=5,
    scoring='r2',
    n_jobs=-1,
    verbose=1
)

# Fit GridSearchCV
grid_search.fit(X_train, y_train)

print("\nGridSearchCV completed!")
print(f"Best parameters: {grid_search.best_params_}")
print(f"Best R² score: {grid_search.best_score_:.4f}")

# Get the best model
best_rf_model = grid_search.best_estimator

# Evaluate on test set
best_test_pred = best_rf_model.predict(X_test)
best_r2 = r2_score(y_test, best_test_pred)
best_mae = mean_absolute_error(y_test, best_test_pred)

print(f"\nTest set performance with best parameters:")
print(f"R² Score: {best_r2:.4f}")
print(f"MAE: ${best_mae:,.2f}")
```
